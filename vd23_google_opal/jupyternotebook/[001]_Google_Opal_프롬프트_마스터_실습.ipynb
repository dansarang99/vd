## 📌 실습 개요
본 노트북은 Google Opal에 접속했을 때 좌측 창에 넣을 구조화된 메타 프롬프트(C-I-P-O-E 규격)를 파이썬 코드로 손쉽게 생성하고, 클립보드에 복사하여 즉시 실습해볼 수 있도록 안내합니다.

### Step 1. 엔진 모듈 임포트 및 환경 점검

```python
import sys
import os
from pathlib import Path

# src 디렉토리 경로 추가
src_path = str(Path(os.getcwd()).parent / "src")
if src_path not in sys.path:
    sys.path.append(src_path)

from opal_prompt_builder import OpalPromptBuilder
print("✅ OpalPromptBuilder 로드 성공!")
```

### Step 2. 창업지도사 5대 프리셋 프롬프트 살펴보기

```python
presets = OpalPromptBuilder.PRESETS
for key, val in presets.items():
    print(f"🔹 [{key}] : {val['name_kr']} ({val['name']})")
```

### Step 3. [실습 1] 원클릭 린 캔버스 프롬프트 생성 & 확인

```python
prompt_lean_canvas = OpalPromptBuilder.generate_preset_prompt("lean_canvas")
print("=" * 60)
print(prompt_lean_canvas)
print("=" * 60)
```

### Step 4. [실습 2] 나만의 맞춤형 자동화 앱 프롬프트 만들기

```python
custom_prompt = OpalPromptBuilder.generate_custom_prompt(
    app_name="AI Startup IR Deck Coach",
    goal="창업자의 5분 피칭 대본을 입력받아 VC 관점의 핵심 전달력 및 숫자 신뢰도를 채점하고 개선안을 제시합니다.",
    target_user="초기 창업팀 및 IR 준비 기업",
    persona="Head of Investment Screening at Tier-1 Accelerator",
    inputs=[
        {"name": "Pitch Script / Outline", "type": "Long Text Area", "desc": "5분 발표 피칭 대본"},
        {"name": "Target Investor Audience", "type": "Dropdown", "options": ["엔젤/AC", "초기 VC", "정부지원사업 심사위원"]}
    ]
)

print("✨ 맞춤형 Opal 프롬프트 완성:")
print(custom_prompt)
```

### Step 5. Google Opal(opal.google)에 접속하여 실행하기
1. 위에서 출력된 프롬프트 텍스트를 복사합니다.
2. 브라우저에서 `https://opal.google` 에 접속합니다.
3. 좌측 입력창에 붙여넣고 `Enter`를 눌러 다단계 노드 그래프가 자동 생성되는 것을 감상하세요!
